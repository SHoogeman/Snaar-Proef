```python
import numpy as np
import glob
import matplotlib.pyplot as plt
from IPython.display import Image
```

# Automatic data overview notebook

You can use this notebook to automatically generate an overview of all the data you have taken of the day. It can show you all scope screenshots, all generator screenshots, and will make plots of all the traces you saved. 

For each file it finds, it will print the file name as text so you can easily copy-paste it into another notebook. 

Note that Jupyter these days "overtakes" the usual right-click mouse menu of your web browser, making it a bit tricker to simply grab a copy of an image from a notebook. However, on Gary's macbook in Google Chrome, it seems that if you use "Shift + Right-click" on an image, then you get back to the built-in Chrome menu and you can then easily use the "Copy Image" option to grab an image and then paste it into your logbook file. 

Below, you should first fill in the name of the date folder (format YYYY-MM-DD) that you want to make the summary of and then run the code to generate the overview. 

```python
folder = "2025-11-13"
```

## Saved Traces Overview

```python
dat_files = glob.glob(folder + "/*.dat")
for f in dat_files:
    data = np.loadtxt(f)
    print("Filename:\n%s" % f)
    t = data[:,0]
    v = data[:,1]
    plt.plot(t,v)
    plt.xlabel("Time (s)")
    plt.ylabel("Voltage (V)")
    plt.show()
```

## Scope screenshot overview

```python
screenshot_files = glob.glob(folder + "/scope*.png")
for f in dat_files:
    print("Filename:\n%s" % f)
    Image(f)
```
