## Example code for Load Line plots

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# observed open-circuit voltage, replace obviously with your own observations :)
V_open = 1006 # V

# observed closed circuit current
I_closed = 0.893e-3 # A

# Guess at Thevinin-Norton equivalant impedance, adjust by hand until it looks reasonable :)
# Reminder: V = IR, R = V/I, I = V/R
# Here, I have made a guess. But you can probably work out a formula and put it in here
# And if you want, you could also fit it using curve_fit() or perform a simple linear regression 
R_th = 1e6

# The load line: vertical axis is current, the horizontal axis is voltage
# https://en.wikipedia.org/wiki/Load_line_(electronics)

V = np.array([0, V_open])
I = np.array([I_closed, 0])

plt.plot(V, I, "o")
plt.xlabel("Voltage (V)")
plt.ylabel("Current (A)")
# Axis lines are useful
plt.axhline(0, ls=":", c='grey')
plt.axvline(0, ls=":", c='grey')

I_load_line = I[0] - V/R_th
plt.plot(V, I_load_line, label="R_th = %e" % R_th)
plt.legend()
plt.title("Load line for (describe your configuration here!)")
plt.show()
```

## Example code for the IV code of a diode

Here, we use the two channels to record, vs time, both V_in and V_out for the "diode divider". 

Say we call the data from trace 1 as (t,x) and then from trace 2 as (t,y). 

By ignoring the time axes of the trace, you can then make an "x-y" plot of V_in vs V_out by using hte 

For this, it is good to make sure the scope is "stopped" so that you don't accidentally download trace 1 from one acquisition and trace 2 from the next acquisition. Although this is pretty unlikely (python is fast), if it does happen, your x-y plot will look like nonsense...

Now grab the scope:

```python
scope = Scope("SDS")
```

Now you can use this to grab the two traces (I have assumed that V_in is on Ch1 and V_out is on channel 2):

```python
t,V_in = scope.get_trace(1)
t,V_out = scope.get_trace(2)
```

```python
plt.plot(V_in, V_out)
plt.ylabel("V_out (V)")
plt.xlabel("V_in (V)")
plt.title("Measurements for (....describe setup....)")
plt.show()
```
